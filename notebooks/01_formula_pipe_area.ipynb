# Formula 01 - Pipe Cross-Section Area

This notebook uses the simplified engine from `flow.engine`.

```python
import pandas as pd
import matplotlib.pyplot as plt

from flow.engine import (
    PipeAreaInput,
    calculate_pipe_areas,
    pipe_area_outputs_to_records,
)
```

```python
inputs = [
    PipeAreaInput(inner_diameter_m=0.5, pipe_id="A", diameter_source="user_input"),
    PipeAreaInput(inner_diameter_m=1.0, pipe_id="B", diameter_source="user_input"),
    PipeAreaInput(inner_diameter_m=1.5, pipe_id="C", diameter_source="user_input"),
    PipeAreaInput(inner_diameter_m=2.0, pipe_id="D", diameter_source="user_input"),
    PipeAreaInput(inner_diameter_m=2.5, pipe_id="E", diameter_source="user_input"),
]

outputs = calculate_pipe_areas(inputs)
records = pipe_area_outputs_to_records(outputs)
df = pd.DataFrame(records)
df
```

```python
plt.figure(figsize=(6, 4))
plt.plot(df["inner_diameter_m"], df["area_m2"], marker="o")
plt.xlabel("Inner diameter (m)")
plt.ylabel("Area (m^2)")
plt.title("Pipe Cross-Section Area - Formula 01")
plt.grid(True, alpha=0.3)
plt.show()
```
