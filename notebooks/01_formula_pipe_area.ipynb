# Formula 01 - Pipe Cross-Section Area

This notebook uses the authoritative engine implementation from `flow.engine`.


```python
import pandas as pd
import matplotlib.pyplot as plt

from flow.engine.geometry.pipe_area import (
    PipeAreaInput,
    PipeCrossSectionAreaFormula,
)
from flow.engine.export.dataframe import results_to_records
```

```python
formula = PipeCrossSectionAreaFormula()

inputs = [
    PipeAreaInput(inner_diameter_m=0.5, pipe_id="A", diameter_source="user_input"),
    PipeAreaInput(inner_diameter_m=1.0, pipe_id="B", diameter_source="user_input"),
    PipeAreaInput(inner_diameter_m=2.0, pipe_id="C", diameter_source="user_input"),
]

results = [formula.evaluate(i) for i in inputs]

records = results_to_records(results)
df = pd.DataFrame(records)

# Flatten the nested `output` dict for plotting.
df_plot = pd.json_normalize(df["output"])[["inner_diameter_m", "area_m2"]]
df_plot
```

```python
plt.figure(figsize=(6, 4))
plt.plot(df_plot["inner_diameter_m"], df_plot["area_m2"], marker="o")
plt.xlabel("Inner diameter (m)")
plt.ylabel("Area (m^2)")
plt.title("Pipe Cross-Section Area - Formula 01")
plt.grid(True, alpha=0.3)
plt.show()
```
