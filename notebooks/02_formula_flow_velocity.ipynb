# Formula 02 - Flow Velocity

This notebook uses the simplified engine from `flow.engine`.

Formula chain:
- Formula 01: inner_diameter_m -> area_m2
- Formula 02: flow_rate + area -> velocity

```python
import pandas as pd
import matplotlib.pyplot as plt
import math

from flow.engine import (
    PipeAreaInput,
    calculate_pipe_area,
    FlowVelocityInput,
    calculate_flow_velocities,
    flow_velocity_outputs_to_records,
)
```

## Calculate area using Formula 01

Then use Formula 01 output as Formula 02 input.

```python
# Calculate area for a 0.1m diameter pipe
area_result = calculate_pipe_area(PipeAreaInput(inner_diameter_m=0.1))
area_m2 = area_result.area_m2
print(f"Pipe area: {area_m2:.6f} m²")
```

## Calculate velocity for several flow rates through the same pipe

```python
flow_rates = [0.001, 0.005, 0.01, 0.02, 0.05, 0.1]

velocity_inputs = [
    FlowVelocityInput(
        volumetric_flow_rate_m3_s=rate,
        area_m2=area_m2,
        flow_id=f"Q-{rate}",
        area_source="formula_01_pipe_area",
    )
    for rate in flow_rates
]

velocity_outputs = calculate_flow_velocities(velocity_inputs)
records = flow_velocity_outputs_to_records(velocity_outputs)
df = pd.DataFrame(records)
df
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df["volumetric_flow_rate_m3_s"], df["velocity_m_s"], marker="o")
plt.xlabel("Flow Rate (m³/s)")
plt.ylabel("Velocity (m/s)")
plt.title("Flow Velocity vs Flow Rate (Fixed Pipe Diameter)")
plt.grid(True, alpha=0.3)
plt.show()
```

## Calculate velocity for several diameters at the same flow rate

```python
flow_rate = 0.01  # m³/s
diameters = [0.05, 0.1, 0.15, 0.2, 0.25, 0.3]

diameter_inputs = []
for d in diameters:
    area = calculate_pipe_area(PipeAreaInput(inner_diameter_m=d))
    diameter_inputs.append(
        FlowVelocityInput(
            volumetric_flow_rate_m3_s=flow_rate,
            area_m2=area.area_m2,
            flow_id=f"D-{d}",
            area_source="formula_01_pipe_area",
        )
    )

diameter_outputs = calculate_flow_velocities(diameter_inputs)
diameter_records = flow_velocity_outputs_to_records(diameter_outputs)
df2 = pd.DataFrame(diameter_records)
df2
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df2["inner_diameter_m"], df2["velocity_m_s"], marker="o")
plt.xlabel("Pipe Inner Diameter (m)")
plt.ylabel("Velocity (m/s)")
plt.title("Flow Velocity vs Pipe Diameter (Fixed Flow Rate)")
plt.grid(True, alpha=0.3)
plt.show()
```
